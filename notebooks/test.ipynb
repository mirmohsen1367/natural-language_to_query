```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("/home/mousavi-m/Documents/talk-to-django/all-MiniLM-L6-v2")
```

```python
docs = ["The dog jumped over the cat", 
        "The cat jumped over the dog", 
        "it is very warm today", 
        "The cat is yellow and the dog is red"
        ]
```

```python
document_embeddings = []
dataset = [
    {"id": 1, "text": "The dog jumped over the cat"},
    {"id": 2, "text": "The cat jumped over the dog"},
    {"id": 3, "text": "it is very warm today"},
    {"id": 4, "text": "it was cold yesterday"},
]
dataset
```

```python
for x in dataset:
    embed = model.encode(x.get("text"))
    final_data = (x.get("id"), embed)
    document_embeddings.append(final_data)
```

```python
query = "it was cold yesterday"
query_embedding = model.encode([query])
```

```python
results = []
for doc in document_embeddings:
    doc_id = doc[0]
    document_embedding = doc[1]
    rank = model.similarity(document_embedding, query_embedding)
    results.append((doc_id, rank))
```

```python
results.sort(key=lambda x: x[1], reverse=True)
```

```python
results
```

```python

for result in results:
    dataset_id = result[0]
    rank = result[1]
    doc = next(doc for doc in dataset if doc["id"] == dataset_id)
    print(dataset_id, rank, doc)
```
